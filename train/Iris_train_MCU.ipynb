```python
# Gerekli kütüphanelerin içe aktarılması
import numpy as np
import tensorflow as tf
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
```

```python
# Iris veri kümesinin yüklenmesi
iris = load_iris()
X = iris.data  # Özellikler: Sepal ve Petal uzunluk/genişlik
y = iris.target.reshape(-1, 1)  # Hedef değişken: Çiçek türleri

# Hedef değişkenin One-Hot Encoding işlemi
encoder = OneHotEncoder(sparse_output=False)
y_encoded = encoder.fit_transform(y)

# Verinin eğitim ve test setlerine bölünmesi
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)
```

```python
# Modelin oluşturulması
model = tf.keras.models.Sequential([
    tf.keras.layers.Dense(10, activation='relu', input_shape=(4,)),
    tf.keras.layers.Dense(10, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])

# Modelin derlenmesi
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# Modelin eğitilmesi
model.fit(X_train, y_train, epochs=100, batch_size=5, verbose=1)
```

```python
# Modelin test verisi üzerindeki performansının değerlendirilmesi
loss, accuracy = model.evaluate(X_test, y_test, verbose=0)
print(f"Test kaybı: {loss:.4f}")
print(f"Test doğruluğu: {accuracy:.4f}")
```

```python
# Yeni giriş verileri
new_samples = np.array([
    [5.1, 3.5, 1.4, 0.2],  # Muhtemelen Setosa
    [6.0, 2.2, 4.0, 1.0],  # Muhtemelen Versicolor
    [6.9, 3.1, 5.4, 2.1],  # Muhtemelen Virginica
    [5.5, 2.5, 4.0, 1.3]   # Muhtemelen Versicolor
])

# Tahminlerin yapılması
predictions = model.predict(new_samples)

# Tahmin sonuçlarının yazdırılması
predicted_classes = np.argmax(predictions, axis=1)
for i, pred in enumerate(predicted_classes):
    species = iris.target_names[pred]
    print(f"Giriş {i+1}: Tahmin edilen tür -> {species}")
```

```python
import tensorflow as tf
import numpy as np

# 1. Adım: .h5 modelinin yolunu belirtin
h5_model_path = "iris.h5"  # Kendi .h5 modelinizin yolunu buraya yazın

# 2. Adım: .h5 modelini yükleyin ve TFLite'a dönüştürün
try:
    # Modeli yükle
    model = tf.keras.models.load_model(h5_model_path)
    print(f".h5 modeli yüklendi: {h5_model_path}")

    # Modeli TFLite'a dönüştür
    converter = tf.lite.TFLiteConverter.from_keras_model(model)
    # converter.target_spec.supported_ops = [tf.lite.OpsSet.TFLITE_BUILTIN]
    tflite_model = converter.convert()

   
    # Eğer modeli bir dosyaya kaydetmek isterseniz:
    with open("model.tflite", "wb") as f:
        f.write(tflite_model)
    print("TFLite modeli dosyaya kaydedildi: model.tflite")

except Exception as e:
    print(f"Hata: İşlem sırasında bir hata oluştu: {e}")
```

```python
def convert_tflite_to_header(tflite_path, header_path, array_name="g_model"):
    with open(tflite_path, "rb") as f:
        model_data = f.read()

    with open(header_path, "w") as f:
        f.write('#include "model.h"\n\n')
        f.write('// clang-format off\n')
        f.write(f'alignas(8) const unsigned char {array_name}[] = {{\n')

        # Her satıra 7 byte
        lines = []
        for i in range(0, len(model_data), 7):
            chunk = model_data[i:i+7]
            line = ", ".join(f"0x{b:02x}" for b in chunk)
            lines.append(f"  {line},")

        # Tüm satırları yaz
        f.write("\n".join(lines))
        f.write('\n};\n')
        f.write('// clang-format on\n\n')
        f.write(f'const int {array_name}_len = {len(model_data)};\n')
```

```python
convert_tflite_to_header("model.tflite", "model.h")
```
